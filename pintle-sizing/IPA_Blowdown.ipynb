```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
IN = 0.0254
PSI = 6894.76
```

```python
# Don't change these for injector sizing
mass_flow = 0.480 # kg/s
density = 786 # kg/m3
roughness = 0.000045
dynamic_viscocity = 0.002381

# Change these for injector sizing
height = 0.3 * IN
width = 0.006 * IN #dont change this
area = (height * width) + (np.pi * (width / 2)**2)
perimeter = 2 * (height + np.pi * width / 2)
hydraulic_diameter = 4 * area / perimeter
length = 0.2 * IN # inches
hole_count = 24

velocity = (mass_flow / hole_count) / (density * area)
reynolds = density * velocity * hydraulic_diameter / dynamic_viscocity
friction_factor = 0.25 * np.log10((roughness / (3.7 * hydraulic_diameter)) + (5.74 / reynolds**0.9))**-2
loss_per_meter = friction_factor * density * velocity**2 / (2 * hydraulic_diameter)
loss = loss_per_meter * length
print(f"Total loss: {loss / PSI} PSI")
print(f"Injector velocity: {velocity} m/s")
```
